```python
# Q1
# i
roll_no = 1024170442
L = [int(digit) * 10 for digit in str(roll_no)]
print("Original L: " , L)

#ii
L.append(55)
print("L after appending 55: " , L)
# append() added 55 at the end of the list.

L.insert(2, 25)
print("L after inserting 25 at index 2: " , L)
# 25 was inserted at index 2, and the existing elements
# from index 2 onwards shifted one position to the right.

#iii
L.remove(10)
print("L after removing 10: " , L)

removed = L.pop(3)
print("L after popping element at index 3: " , L)
print("Removed element: " , removed)

#iv
L.sort()
print("L after sorting in ascending order: " , L)
L.sort(reverse=True)
print("L after sorting in descending order: " , L)

#v
print("First 3 elements: " , L[:3])
print("Last 3 elements: " , L[-3:])

#vi
average = sum(L) / len(L)
greater_than_average = [x for x in L if x>average]

print("Average of L: " , average)
print("Elements greater than average: " , greater_than_average)
```

```python
# Q2
# i
scores = (10, 0, 20, 40, 10, 70, 0, 40)
print("Scores: " , scores)
maximum = max(scores)
maximum_index = scores.index(maximum)
minimum = min(scores)
minicount = scores.count(minimum)
print("Highest score: " , maximum)
print("Index of highest score: " , maximum_index)
print("Lowest score: " , minimum)
print("Count of lowest score: " , minicount)

#ii
reversed_scores = list(reversed(scores))
print("Reversed scores: " , reversed_scores)
# Tuples are immutable, so their elements cannot be changed directly.

#iii
score = int(input("Enter a score to check: "))

if score in scores:
    print(f"The score is present at index {scores.index(score)}")
else:
    print(f"The score {score} is not present in the tuple.")
  
#v
first_score, second_score, *remaining_scores = scores
print("First score: " , first_score)
print("Second score: " , second_score)
print("Remaining scores: " , remaining_scores)

#iv
scores[0] = 100
# This will raise a TypeError because tuples are immutable and do not support item assignment.
```

```python
import random
random.seed(1024170442)

#i
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random numbers: " , numbers)

#ii
odd_numbers = [x for x in numbers if x%2!=0]
print("Odd numbers: " , odd_numbers)
print("Number of odd numbers: " , len(odd_numbers))

#iii
even_numbers = [x for x in numbers if x%2==0]
print("Even numbers: " , even_numbers)
print("Number of even numbers: " , len(even_numbers))

#iv
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime numbers: " , prime_numbers)
print("Number of prime numbers: " , len(prime_numbers))

#v
frequency = {}
for x in numbers:
    if x in frequency:
        frequency[x] += 1
    else:
        frequency[x] = 1

most_frequent = max(frequency, key=frequency.get)
print("Most frequent number: " , most_frequent)
print("Frequency of most frequent number: " , frequency[most_frequent])
```

```python
#Q4
setA = {int(digit*7) for digit in str(roll_no)}
setB = {int(digit*9) for digit in str(roll_no)}

print("Set A: " , setA)
print("Set B: " , setB)

#i
union_set = setA.union(setB)
print("Union of Set A and Set B: " , union_set)

#ii
intersection_set = setA.intersection(setB)
print("Intersection of Set A and Set B: " , intersection_set)

#iii
difference_setAB = setA.difference(setB)
print("Difference of Set A and Set B (A - B): " , difference_setAB)

difference_setBA = setB.difference(setA)
print("Difference of Set B and Set A (B - A): " , difference_setBA)
#difference() is one-directional, while symmetric_difference() combines both differences.

#iv
symmetric_difference_set = setA.symmetric_difference(setB)
print("Symmetric difference of Set A and Set B: " , symmetric_difference_set)

#v
is_subset = setA.issubset(setB)
is_superset = setB.issuperset(setA)
print("Is Set A a subset of Set B? " , is_subset)
print("Is Set B a superset of Set A? " , is_superset)

#vi
X = int(input("Enter a value to remove from Set A: "))
setA.discard(X)
print("Set A after removing X: " , setA)
# discard() is safer than remove() because it doesn't raise an error if the value is not found in the set.  
```

```python
#Q5
my_dict = {
    "name": "Mayank Garg",
    "roll_no": 1024170442,
    "branch": "Computer Science",
    "age": 20,
    "city": "Patiala"
}
#i
my_dict["location"] = my_dict.pop("city")

#ii
my_dict["cgpa"] = 9.5

#iii
my_dict["age"] = my_dict.get("age", 0) + 1

#iv
my_dict_copy1 = my_dict.copy()
removed_branch = my_dict_copy1.pop("branch", None)
print("Removed branch using pop(): " , removed_branch)
print("Dictionary after pop(): " , my_dict_copy1)


my_dict_copy2 = my_dict.copy()
del my_dict_copy2["branch"]  
print("Dictionary after del(): " , my_dict_copy2)
#del doesn't return anything while the pop() method returns the value of the removed key.

#v
for key,value in my_dict.items():
    print(f"{key} → {value}")

#vi
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key does not exist in the dictionary.")


list(my_dict.keys())
list(my_dict.values())

#vii
friend_dict = {
    "name": "Paryag Bansal",
    "roll_no": 1024170446,
    "age": 20,
    "branch": "Electrical Engineering",
    "city": "Chandigarh"
}
merged_dict = {**my_dict, **friend_dict}
print("Merged Dictionary: ", merged_dict)
#When both dictionaries contain the same key, the value from the rightmost/second dictionary (friend_dict) wins.

#viii
string_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print("Dictionary with only string values: ")
print(string_dict)
```
