## Q1. Basic operations on a 1-D NumPy array

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

# a) Add 2 to every element
print("After adding 2:", arr + 2)

# b) Multiply every element by 3
print("After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("After dividing by 2:", arr / 2)
```

## Q2. Basic NumPy array operations

### Q2(a). Reverse the given NumPy array

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

### Q2(b). Find the most frequent value and its indices

```python
# Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent_x = values[np.argmax(counts)]
indices_x = np.where(x == most_frequent_x)[0]

print("Array x:", x)
print("Most frequent value:", most_frequent_x)
print("Frequency:", np.max(counts))
print("Indices:", indices_x)

# Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent_y = values[np.argmax(counts)]
indices_y = np.where(y == most_frequent_y)[0]

print("\nArray y:", y)
print("Most frequent value:", most_frequent_y)
print("Frequency:", np.max(counts))
print("Indices:", indices_y)
```

## Q3. Access elements using row and column indices

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("2-D array:")
print(arr)

# a) 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

## Q4. `linspace()`, array properties, reshape and transpose

```python
# 25 evenly spaced numbers from 10 to 100
Mayank = np.linspace(10, 100, 25)

print("Array:")
print(Mayank)

print("\nDimensions:", Mayank.ndim)
print("Shape:", Mayank.shape)
print("Total elements:", Mayank.size)
print("Data type:", Mayank.dtype)
print("Total bytes consumed:", Mayank.nbytes)

# Transpose using reshape()
transpose_reshape = Mayank.reshape(1, 25)
print("\nTranspose using reshape():")
print(transpose_reshape)

# Transpose using T
transpose_T = Mayank.T
print("\nUsing T attribute:")
print(transpose_T)

print("\nNote: For a 1-D NumPy array, T does not change its shape.")
print("To get a 25 x 1 transpose-like column array, use Mayank.reshape(25, 1).")
```

## Q5. 2-D array statistics, reshape and resize

```python
ucs420_Mayank = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Mayank)

# Statistical operations
print("\nMean:", np.mean(ucs420_Mayank))
print("Median:", np.median(ucs420_Mayank))
print("Maximum:", np.max(ucs420_Mayank))
print("Minimum:", np.min(ucs420_Mayank))
print("Unique elements:", np.unique(ucs420_Mayank))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Mayank = ucs420_Mayank.reshape(4, 3)

print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_Mayank)

# Resize to 2 rows and 3 columns
resized_ucs420_Mayank = np.resize(ucs420_Mayank, (2, 3))

print("\nResized array (2 x 3):")
print(resized_ucs420_Mayank)
```
